```python
import pandas as pd
```

```python
# 1. Load the dataset
df = pd.read_csv("Sample - Superstore.csv", encoding="windows-1252")
print("Original Shape:", df.shape)
```

```python
# 2. Drop unnecessary columns
# 'Row ID' is not required for the analysis, so we remove it
if 'Row ID' in df.columns:
    df.drop('Row ID', axis=1, inplace=True)
```

```python
# 3. Handle duplicate rows
duplicates = df.duplicated().sum()
print(f"Total Duplicate Rows: {duplicates}")
if duplicates > 0:
    df.drop_duplicates(inplace=True)
```

```python
# 4. Correct data types by converting date columns to datetime format
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
# 5. Feature Engineering
# Create two new columns to derive additional business insights

# a) Delivery Days: Calculate the number of days between order and shipment
df['Delivery_Days'] = (df['Ship Date'] - df['Order Date']).dt.days

# b) Profit Margin: Calculate the profit percentage for each transaction
# Formula: (Profit / Sales) * 100
df['Profit_Margin_%'] = (df['Profit'] / df['Sales']) * 100
print("\nCleaning Completed. Final Shape:", df.shape)
```

```python
# 6. Export the cleaned dataset
df.to_csv("Cleaned_Superstore_Sales.csv", index=False)
print("Cleaned dataset successfully saved as 'Cleaned_Superstore_Sales.csv'")
```
